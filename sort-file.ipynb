# Sorting a file with Python

In this section, we create a function `sortFile(filename)` which takes the name of a file, reads all of the lines in that file and sorts them.

## sortFile(filename)

This function should return a list containing the lines of the file in sorted order. For example, if the passed in file contained
```
xyz
abc
def
qqq
```

This function would return the list:

```
['abc', 'def','qqq','xyz']
```

Remember that \n representation of the newline character (the character that appears at the end of each line to tell the computer to
start printing on the next line).  If you were to print a list with strings that have newlines in them, Python prints this special
representation.  If you just print the string by itself, it prints the actual newline character to go to the next line.  You don't need to do anything special about this, we just did not want you to be confused about what we wrote above or anything you might see if you need to
debug your code.

Note that you can (and should!) use the built-in sorting functions to sort the list.  There are two different functions, which work in slightly different ways.

One is the `sorted()` function which takes a list, and returns a new list with the values sorted.  The original list is unchanged:
```
x = [5, 3, 1, 2, 4]
y = sorted(x)
print(x)
print(y)
```
This will print: 
```
[5, 3, 1, 2, 4]
[1, 2, 3, 4, 5]
```
because x is unchanged by the call to sorted, but returns `[1, 2, 3, 4, 5]` which is assigned to y.

The other way to sort a list is to call the `.sort()` method on the list, which modifies the list (and returns `None`), so
```
x = [5, 3, 1, 2, 4]
y = x.sort()
print(x)
print(y)
```
Would print:
```
[1, 2, 3, 4, 5]
None
```
because x got modified to be in sorted order, and the call to `sort()` returned None.  Note that we typically do not assign the `return` value of `x.sort()` to anything, as it is always None, so is not terribly useful. We only did it here for the purposes of illustration.

Write the function `sortFile(filename)` and read all lines from the file `input.txt`, sort them alphabetically, and return the sorted list of lines.

````python
def sortFile(filename):
    # Replace this with your code

    # Open a file and guarantee that it automatically closes
    with open (filename, "r") as file:
        # Read all the lines of a file and returns them as a list of strings
        lines = file.readlines()

    # Sort the list alphabetically
    lines.sort()

    # Return the sorted list
    return lines
````
