## "Simplified" Roman Numerals

Roman numerals are a system of writing numbers used by the ancient Romans. The system involves letter symbols that represent numbers. This simplified Roman numerals will have the same symbols as real Roman numerals.

### Question 1.1

The letter symbols used in Roman numerals and the corresponding numerical values they represent are listed in Table 1.1.: 

_**Table 1.1**. Basic Roman numeral system._

| Symbol | Value |
|:------:|:--------------:|
| `M`    | `1000`            |
| `D`    | `500`            |
| `C`    | `100`            | 
| `L`    | `50`            |
| `X`    | `10`            |
| `V`    | `5`            |
| `I`    | `1`            |

However, to keep things simple, our Roman numerals will be purely additive. Real Roman numerals allow for "subtractive forms" where a smaller number before a bigger number means to subtract, such as IV means four *`(V - I = 5 - 1 = 4)`*. To simplify your task, we are going to only write additions, from largest to smallest, so we will write four as *`III (1 + 1 + 1 + 1)`*.

Your goal is to take a positive number and return a string that is the simplified Roman numeral representation. Note that the largest numbers should be first (Ms before Ds, Ds before Cs, and so on). The answer should make the best use of the biggest numbers (so 5 should be V, not IIIII).

For example, `roman_numeral(2841)` would return `MMDCCCXXXXI` because

2841 = 1000 + 1000 + 500 + 100 + 100 + 100 + 10 + 10 + 10 + 10 + 1

     M      M      D     C     C     C    X    X    X    X    I

Note that our function only needs to deal with positive numbers, we will nor test on negative numbers nor zero.

````python
def roman_numeral(n):
    ans = ""
    
    # Write your code here

    values = [1000, 500, 100, 50, 10, 5, 1]
    symbols = ["M", "D", "C", "L", "X", "V", "I"]

    for i in range(len(values)):
        while n >= values[i]:
            ans = ans + symbols[i]
            n = n - values[i]
    return ans

for i in range(1, 25):
    print(roman_numeral(i))

print(roman_numeral(50))
print(roman_numeral(100))
print(roman_numeral(500))
print(roman_numeral(1000))
print(roman_numeral(178))
print(roman_numeral(2841))
print(roman_numeral(8492))
````
